```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

df = pd.read_csv('../data/car_prices.csv')

print(df.dtypes)
print("Prezzi mancanti totali:", df['Price'].isna().sum())
```

```python
prezzi = pd.Series(df.Price)

plt.Figure(figsize=(10,4))

sns.boxplot(x=prezzi,color='salmon')
plt.xlabel('Prezzi ¢')
plt.title('Outlier nei Prezzi')
plt.show()
```
